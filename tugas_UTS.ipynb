````python
# import Library
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

RANDOM_STATE = 76   # isi dengan 2 digit terakhir NIM Anda
````

````python
# Salin fungsi train_test_split dari materi di sini
def train_test_split(X, y, test_size=0.2, random_state=None):
    if random_state is not None:
        np.random.seed(random_state)
    n_samples = len(X)
    indices = np.arange(n_samples)
    np.random.shuffle(indices)
    
    test_len = int(n_samples * test_size)
    test_idx = indices[:test_len]
    train_idx = indices[test_len:]
    
    return X[train_idx], X[test_idx], y[train_idx], y[test_idx]
````

### Soal A3 – Training dengan Normal Equation
1. Latih model `LinearRegression` (normal equation, seperti di materi) menggunakan data training.
2. Tampilkan nilai **intercept** ($b$) dan **koefisien** ($m$), lalu tuliskan persamaan garis regresinya: $\hat{y} = b + mX$.
3. **Interpretasikan** arti nilai $m$ dalam konteks kasus ini (misal: "setiap kenaikan biaya iklan 1 juta Rp, penjualan naik sebesar ...").

````python
# Jawaban A3
class LinearRegressionNormal:
    def fit(self, X, y):
        X_b = np.c_[np.ones((X.shape[0], 1)), X]
        theta = np.linalg.inv(X_b.T.dot(X_b)).dot(X_b.T).dot(y)
        self.intercept_ = theta[0]
        self.coef_ = theta[1:]
        
    def predict(self, X):
        X_b = np.c_[np.ones((X.shape[0], 1)), X]
        return X_b.dot(np.r_[[self.intercept_], self.coef_])

model_a = LinearRegressionNormal()
model_a.fit(X_train_a, y_train_a)

b = model_a.intercept_
m = model_a.coef_[0]

print(f"Intercept (b): {b:.4f}")
print(f"Koefisien (m): {m:.4f}")
print(f"Persamaan Garis Regresi: y_hat = {b:.4f} + {m:.4f} * X")
````

### Soal A5 – Gradient Descent & Learning Rate
1. Latih model `LinearRegressionGD` (dari materi) dengan `n_iterations=5000` menggunakan **3 nilai learning rate**: `0.001`, `0.02`, dan `0.1`.
2. Untuk setiap learning rate, tampilkan intercept dan koefisien akhirnya, lalu bandingkan dengan hasil normal equation (Soal A3).
3. Buat grafik **loss (MSE) vs iterasi** untuk ketiga learning rate (boleh dalam satu grafik atau *subplot*; gunakan skala log jika perlu).
4. Jelaskan apa yang terjadi pada masing-masing learning rate: mana yang **terlalu lambat**, mana yang **pas**, dan mana yang **gagal konvergen (divergen)**? Mengapa bisa terjadi?

> *Tips: jika muncul `RuntimeWarning: overflow`, itu adalah bagian dari hasil eksperimen. Jelaskan artinya.*

````python
# Jawaban A5
class LinearRegressionGD:
    def __init__(self, learning_rate=0.01, n_iterations=1000):
        self.lr = learning_rate
        self.n_iter = n_iterations
        self.loss_history = []
        
    def fit(self, X, y):
        n = X.shape[0]
        self.theta = np.zeros(X.shape[1] + 1)
        X_b = np.c_[np.ones((n, 1)), X]
        
        for _ in range(self.n_iter):
            predictions = X_b.dot(self.theta)
            errors = predictions - y
            loss = np.mean(errors ** 2)
            self.loss_history.append(loss)
            
            gradients = (2/n) * X_b.T.dot(errors)
            self.theta -= self.lr * gradients
            
        self.intercept_ = self.theta[0]
        self.coef_ = self.theta[1:]

learning_rates = [0.001, 0.02, 0.1]
plt.figure(figsize=(10, 6))

for lr in learning_rates:
    gd_model = LinearRegressionGD(learning_rate=lr, n_iterations=5000)
    gd_model.fit(X_train_a, y_train_a)
    print(f"LR: {lr} -> Intercept: {gd_model.intercept_:.4f}, Coef: {gd_model.coef_[0]:.4f}")
    plt.plot(gd_model.loss_history, label=f'LR = {lr}')

plt.yscale('log')
plt.title('Loss (MSE) vs Iterasi untuk Berbagai Learning Rate')
plt.xlabel('Iterasi')
plt.ylabel('MSE Loss (Skala Log)')
plt.legend()
plt.grid(True)
plt.show()
````

**Jawaban A5 (analisis):**
1. Learning Rate 0.001 (Terlalu Lambat): Penurunan loss (MSE) sangat landai dan lambat karena langkah pembaruan bobot terlalu kecil, sehingga membutuhkan iterasi yang jauh lebih banyak untuk mencapai konvergensi.
2. Learning Rate 0.02 (Pas/Optimal): Kurva loss menurun secara mulus dan stabil menuju titik minimum, menghasilkan bobot akhir yang sangat mendekati hasil normal equation.
3. Learning Rate 0.1 (Terlalu Besar/Divergen): Ukuran langkah pembaruan terlalu besar sehingga melewati titik minimum (overshooting), yang menyebabkan loss melonjak naik tak beraturan atau memicu RuntimeWarning: overflow.

### Soal B2 – Split Data & Training
1. Bagi data dengan `test_size=0.25` dan `random_state=RANDOM_STATE`.
2. Latih model `LogisticRegression` (gradient descent, dari materi) dengan `learning_rate=0.1` dan `n_iterations=10000`.
3. Tampilkan intercept ($b$) dan koefisien ($m_1, m_2$), lalu tuliskan persamaannya:
   $$P(\text{Spam}) = \sigma(b + m_1 \cdot \text{JumlahLink} + m_2 \cdot \text{KataPromo})$$
4. Buat grafik **log loss vs iterasi**. Apakah loss sudah turun dan stabil?

````python
# Jawaban B2
X_train_b, X_test_b, y_train_b, y_test_b = train_test_split(X_email, y_email, test_size=0.25, random_state=RANDOM_STATE)

class LogisticRegressionGD:
    def __init__(self, learning_rate=0.1, n_iterations=10000):
        self.lr = learning_rate
        self.n_iter = n_iterations
        self.loss_history = []
        
    def sigmoid(self, z):
        return 1 / (1 + np.exp(-np.clip(z, -500, 500)))
        
    def fit(self, X, y):
        n, m = X.shape
        self.theta = np.zeros(m + 1)
        X_b = np.c_[np.ones((n, 1)), X]
        
        for _ in range(self.n_iter):
            z = X_b.dot(self.theta)
            h = self.sigmoid(z)
            epsilon = 1e-5
            loss = -np.mean(y * np.log(h + epsilon) + (1 - y) * np.log(1 - h + epsilon))
            self.loss_history.append(loss)
            
            gradient = (1 / n) * X_b.T.dot(h - y)
            self.theta -= self.lr * gradient
            
        self.intercept_ = self.theta[0]
        self.coef_ = self.theta[1:]
        
    def predict_proba(self, X):
        X_b = np.c_[np.ones((X.shape[0], 1)), X]
        return self.sigmoid(X_b.dot(self.theta))
    
    def predict(self, X, threshold=0.5):
        return (self.predict_proba(X) >= threshold).astype(int)

model_b = LogisticRegressionGD(learning_rate=0.1, n_iterations=10000)
model_b.fit(X_train_b, y_train_b)

print(f"Intercept (b): {model_b.intercept_:.4f}")
print(f"Koefisien (m1, m2): {model_b.coef_[0]:.4f}, {model_b.coef_[1]:.4f}")

plt.figure(figsize=(8, 5))
plt.plot(model_b.loss_history, color='purple')
plt.title('Log Loss vs Iterasi (Logistic Regression)')
plt.xlabel('Iterasi')
plt.ylabel('Log Loss')
plt.grid(True)
plt.show()
````

### Soal B3 – Evaluasi Model
Selain `accuracy_score` dan `confusion_matrix` dari materi, **implementasikan sendiri** fungsi berikut:

$$
Precision = \frac{TP}{TP + FP} \qquad
Recall = \frac{TP}{TP + FN} \qquad
F1 = 2 \cdot \frac{Precision \cdot Recall}{Precision + Recall}
$$

1. Hitung accuracy, confusion matrix, precision, recall, dan F1-score pada **data testing**.
2. Jelaskan arti nilai **FP** (*false positive*) dan **FN** (*false negative*) pada kasus filter spam. Kesalahan mana yang menurut Anda **lebih merugikan pengguna email**? Mengapa?

````python
# Jawaban 
def confusion_matrix_custom(y_true, y_pred):
    tp = np.sum((y_true == 1) & (y_pred == 1))
    tn = np.sum((y_true == 0) & (y_pred == 0))
    fp = np.sum((y_true == 0) & (y_pred == 1))
    fn = np.sum((y_true == 1) & (y_pred == 0))
    return np.array([[tn, fp], [fn, tp]])

def evaluate_classification(y_true, y_pred):
    cm = confusion_matrix_custom(y_true, y_pred)
    tn, fp, fn, tp = cm[0,0], cm[0,1], cm[1,0], cm[1,1]
    
    acc = (tp + tn) / len(y_true)
    prec = tp / (tp + fp) if (tp + fp) > 0 else 0
    rec = tp / (tp + fn) if (tp + fn) > 0 else 0
    f1 = 2 * (prec * rec) / (prec + rec) if (prec + rec) > 0 else 0
    return acc, cm, prec, rec, f1

y_pred_b3 = model_b.predict(X_test_b, threshold=0.5)
acc, cm, prec, rec, f1 = evaluate_classification(y_test_b, y_pred_b3)

print(f"Accuracy  : {acc:.4f}")
print(f"Precision : {prec:.4f}")
print(f"Recall    : {rec:.4f}")
print(f"F1-Score  : {f1:.4f}")
print(f"Confusion Matrix:\n{cm}")
````

**Jawaban B3 (analisis):**
1. False Positive (FP): Email sah/penting (Bukan Spam) yang salah teridentifikasi oleh sistem sebagai Spam.
2. False Negative (FN): Email Spam yang gagal dideteksi dan lolos masuk ke Kotak Masuk (Inbox) biasa.
3. Kesalahan Paling Merugikan: False Positive (FP) jauh lebih membahayakan pengguna. Jika terjadi FP, pesan krusial seperti konfirmasi pekerjaan atau surat kedinasan akan langsung terbuang ke folder Spam dan berisiko terlewat tanpa disadari. Sementara pada FN, pengguna hanya perlu menghapus email iklan secara manual.
